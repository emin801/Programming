# Task 1

```python
score=int(input("Enter student's score: (0-100)"))
if score>100 or score<0:
    print("Input correct value!")
elif score>91:
    print("Grade is A")
elif score>81:
    print("Grade is B")
elif score>71:
    print("Grade is C")
elif score>61:
    print("Grade is D")
elif score>51:
    print("Grade is E")
else:
    print("Grade is F")
```
